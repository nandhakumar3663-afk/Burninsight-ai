```python
iris=sns.load_dataset('iris')
iris_g=sns.PairGrid(iris)

iris_g.map_upper(sns.kdeplot)
iris_g.map_lower(sns.scatterplot)
iris_g.map_diag(sns.histplot)
iris.head()
```

```python
sns.pairplot(iris)
```

```python
iris=sns.load_dataset('iris')
iris_g=sns.PairGrid(iris,hue='species',
                    x_vars=['petal_length','sepal_length'],
                    y_vars=['petal_width','sepal_width'])

iris_g.map(sns.scatterplot)
iris_g.add_legend()

```
