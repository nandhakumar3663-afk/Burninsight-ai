```python
student_name=input("enter your name:")
age=int(input("enter your age:"))
department=input("enter your department:")
cgpa=float(input("enter your cgpa:"))
internship=bool(input("is internship completed(True/False):"))

print("Name:",student_name,"Type:",type(student_name))
print("Age:",age,"Type:",type(age))
print("Department:",department,"Type:",type(department))
print("Cgpa:",cgpa,"Type:",type(cgpa))
print("Internship Completion Status:",internship,"Type:",type(internship))
```
